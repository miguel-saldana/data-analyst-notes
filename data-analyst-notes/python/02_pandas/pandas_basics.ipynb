## Estructure

- **Series** --> como una comuna de un libro excel, siempre tienen un indice y pueden tener una cabecera. Pueden tener datos de tipo entero, decimal, texto, etc. pero nunca convinado dos tipos de datos en una sola columna (series) 
- **DataFrame** --> es un conjunto de series similiar a un libro de excel. para almacernalos en una variable se usa `df` 

Las filas son  autoidentificadas como las listas de python se inicia por 0 (0,1,2,3,4...)

## Pricipales funciones

1. **.head()**   --> mostra le PRIME 5 righe del DataFrame
2. **.tail()**   --> mostra le ULTIME 5 righe del DataFrame
3. **.descrive** --> mostra le stadistiche di base del DataFrame (media, min, max, std)
4. **.info()**   --> struttura del database (tipi, null, colonne)
5. **.isna()/isnull()**  --> identifica valori mancanti
6. **.dropna()** --> minuove righe con valori nulli

    `df_clean = df.dropna()`
7. **.fillna()** --> riempie i valori nulli

    `df["ritardo_min"] = df["ritardo_min"].fillna(0)`
8. **.loc[]** --> selezione per etichette / IDENTIFICATTORI

    `df.loc[df["ritardo_min"] > 5]`
9. **.iloc[]** --> selezione per posizione con il INDICE

    `df.iloc[0:2]`
10. **sort_values()** --> ordina i dati

    `df.sort_values(by="ritardo_min", ascending=False)`
11. **gorupby()** --> aggregazione dati

        `df = pd.DataFrame({
        "linea": ["MI-BS", "MI-BS", "MI-TO"],
        "ritardo_min": [5, 10, 0]
        })
        df.groupby("linea")["ritardo_min"].mean()`
12. **value_counts()** --> conteggio di valori

    `df["linea"].value_counts()`
13. **rename()** --> Rinomina colonne

    `df.rename(columns={"ritardo_min": "delay_min"})`
14. **merge()** --> join tra dataset
15. **apply()** --> applica funzione custom
16. **to_csv() / read_csv()** --> transforma el DataFrame en un archivo .csv / lee un archivo .csv para convertirlo en un DataFrame
