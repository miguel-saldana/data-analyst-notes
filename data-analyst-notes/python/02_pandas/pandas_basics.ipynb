## Estructure

- **Series** --> como una comuna de un libro excel, siempre tienen un indice y pueden tener una cabecera. Pueden tener datos de tipo entero, decimal, texto, etc. pero nunca convinado dos tipos de datos en una sola columna (series) 
- **DataFrame** --> es un conjunto de series similiar a un libro de excel. para almacernalos en una variable se usa `df` 

Las filas son  autoidentificadas como las listas de python se inicia por 0 (0,1,2,3,4...)
