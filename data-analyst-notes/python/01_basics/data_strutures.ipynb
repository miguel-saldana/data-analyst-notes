# Le strutture dati di Python

| Data Type  | Ordered | Mutable | Allows Duplicates | Indexing | Key-Value Structure | Syntax Example |
|-----------|--------|--------|------------------|----------|--------------------|----------------|
| List      | Yes    | Yes    | Yes              | Yes      | No                  | [1, 2, 3]      |
| Tuple     | Yes    | No     | Yes              | Yes      | No                 | (1, 2, 3)      |
| Dictionary| Yes*   | Yes    | No (keys)        | No       | Yes                | {"a": 1, "b": 2} |
| Set       | No     | Yes    | No               | No       | No                 | {1, 2, 3}      |

> *Dictionaries are ordered as of Python 3.7+, but they are not indexed like lists or tuples.

## LIST

```python
# LIST → struttura ordinata e modificabile (utile per sequenze operative)

# Example: list of railway maintenance tasks
maintenance_tasks = ["track inspection", "ballast tamping", "rail alignment"]

print(maintenance_tasks)

# accesso al primo elemento
print(maintenance_tasks[0])

# aggiungere un nuovo task
maintenance_tasks.append("overhead line check")
print(maintenance_tasks)

# lunghezza della lista
print(len(maintenance_tasks))
```

## TUPLE

```python
# TUPLE → struttura ordinata ma IMMUTABILE (dati fissi)

# Example: railway coordinates (latitude, longitude)
track_position = (45.4642, 9.1900)

print(track_position)

# accesso ai valori
print(track_position[0])

# utile per dati che non devono cambiare (es: coordinate, ID fissi)
```

## SET

```python
# SET → collezione non ordinata senza duplicati

# Example: unique railway machines used in a project
machines = {"tamping machine", "ballast regulator", "tamping machine"}

print(machines)

# aggiungere una macchina
machines.add("rail grinder")
print(machines)

# verifica presenza
print("tamping machine" in machines)
```

## DICTIONARY

```python
# DICTIONARY → struttura chiave-valore (fondamentale per data analysis)

# Example: railway project information
project = {
    "name": "Milan-Verona Upgrade",
    "length_km": 120,
    "status": "in progress"
}

print(project)

# accesso tramite chiave
print(project["name"])

# aggiungere un nuovo campo
project["budget_million_eur"] = 250
print(project)
```

## REALISTICS DATA ANALYST EXAMPLE

```python
# esempio reale: lista di progetti ferroviari (tipico scenario data analyst)

projects = [
    {"name": "Line A Upgrade", "km": 50, "status": "completed"},
    {"name": "High Speed Line", "km": 200, "status": "in progress"},
    {"name": "Metro Extension", "km": 30, "status": "planned"}
]

# estrarre solo i progetti in corso
in_progress_projects = []

for p in projects:
    if p["status"] == "in progress":
        in_progress_projects.append(p)

print(in_progress_projects)
```
