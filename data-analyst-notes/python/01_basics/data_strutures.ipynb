# LIST

```python
# LIST → struttura ordinata e modificabile (utile per sequenze operative)

# Example: list of railway maintenance tasks -> task = tarea
maintenance_tasks = ["track inspection", "ballast tamping", "rail alignment"]

print(maintenance_tasks)

# accesso al primo elemento
print(maintenance_tasks[0])

# aggiungere un nuovo task con il metodo .append()
maintenance_tasks.append("overhead line check")
print(maintenance_tasks)

# lunghezza della lista
print(len(maintenance_tasks))
```

> Alcuni metodi comuni dell`oggeto lista sono:
- .append() ---> agerga un elemento a la list
- .index() ---> restituye la posicion de un elemento
- .insert() ---> inserta un objeto en una poscion especifica
- .count() ---> cuantas veces un elemento aparece en la list (porque se permiten los duplicados)
- .len() ---> cuenta los elementos de una lista
- .remove() ---> remueve un elemento especifico
- .pop() ---> remueve o sustituye un elemento por default el ultimo (solucion lenta, mejor usar 'deque')
- .sort() ---> ordena la lista de forma creciente (modifica la lista original)
- .reverse() ---> invierte el orde de los elementos
- .copy() ---> crea una copia de la lista
- .clear() ---> vacia completamente la lista

> len() restituisce il numero di elementi in una struttura dati (lista, stringa, ecc.)

Il uso tipico
- contare record
- validare dimensioni dataset
- controlli logici (es. lista vuota)

```python
# lista di tratte ferroviarie
railway_lines = ["Milano-Brescia", "Brescia-Verona", "Verona-Vicenza"]

# numero di tratte
num_lines = len(railway_lines)  # conta gli elementi della lista

print(num_lines)  # output: 3
```

```python
my_list = []
my_list.append('pay bills')
my_list.append('tidy up')
my_list.append('walk the dog')
my_list.append('cook dinner')
print(my_list)
```

```python
# si quisiera agregar un elmento a una posicion puntual dentro de mi lista, primero debo utilizar el metodo .index() para determinar el inidice el elemento antes del que quiero instertar el nuevo elemento
i=my_list.index('cook dinner')
# esto se conbertira en el indice de la nueva actividad que ahora puede ser agregada utilizando el metodo .insert()
my_list.insert(i, 'go to the pharmacy')
print(my_list)
```

## LISTA COME CODA (deque)
usar collections.deque, es mas optimizado per code FIFO, uso 'deque' quando:
- ho molete 'insert'/'remove' all'inizio
- lavoro son stream di dati ()
- gestisco 'code di eventi' (logs, sensori, interventi)


> log = registro cronologico di enventi generati da un sistema. sono la materia prima dei dati e arrivano **in tempo reale**

> quando uso `pandas` no uso `deque`

> `FIFO` = first-in, first-out   `LIFO` = last-in, first-out

- servono per:
    - monitoraggio sistema
    - analisis problemi
    - clacolo KPI
    - AUDIT operativo

- esempi di log reali
    - passaggio treno su una tratta
    - ritardo registrato
    - errore segnalamento
    - intervento manutenzione
    - stato macchinario

> 🧠 Sintesi (mentalità senior)

- log = flusso continuo di eventi
- deque = gestione efficiente del flusso
- pandas = analisi del flusso

```python
# log eventi ferroviari (simulazione)
logs = [
    {"timestamp": "2026-04-10 08:00", "event": "Train passed", "line": "Milano-Brescia"},  # passaggio treno
    {"timestamp": "2026-04-10 08:05", "event": "Delay detected", "delay": 5},             # ritardo
    {"timestamp": "2026-04-10 08:10", "event": "Signal failure"}                          # guasto
]

print(logs)
```

### Coda FIFO (più comune)

```python
from collections import deque  # importo deque

# coda interventi ferroviari
queue = deque()

queue.append("Inspection Milano-Brescia")   # aggiunge in fondo
queue.append("Maintenance Brescia-Verona")

task = queue.popleft()  # rimuove il primo elemento (O(1))

print(task)
```

### Buffer di envti (streaming)

```python
from collections import deque #importo deque

# ultimi 3 eventi di linea
events = deque(maxlen=3) # buffer limitato

events.append('train delay 5 min') # aggiungo evento
events.append('siganl failure') # aggiungo envento
events.append('track inspection') # aggiungo evento
events.append('maintenance complete') # si come ho messo un max di 3 allora il primo vienen eliminato automaticamente

print(events)
```

### Stack + Queue ibrido

```python
from collections import deque # importo deque

tasks = deque()

tasks.append('task a') # push
tasks.append('task b')

tasks.pop() # LIFO
tasks.popleft() # LIFO
```

> 🧠 Mental model (importante)

- `deque` = prima dei dati (stream / pipeline)
- `pandas` = dopo (analisi / reporting)

## LISTA COME PILA (stack)

Una pila implementa la metodologia `LIFO` in cui l'ultimo elemento aggiunto e il primo elemento recuperato.
- `.append()` --> push (inserire)
- `.pop()` --> pop (rimuovere)

```python
# pila di interventi ferrroviari urgenti
stack = []

# con push aggiungo interventi
stack.append('inspection milano-brescia') # inserisco intervento
stack.append('signal check verona') # inserisco intervento
stack.append('track repari vincenza') # inserisco intervento

# pop (il ultimo intervento viene eseguito per primo)
last_tastk = stack.pop() # rimuovo il ultimo elemento (LIFO)

print(last_tastk)
```

> 🧠 Quando usarlo

quando gestisci:
- operazioni urgenti
- undo/redo
- processi a priorità immediata

# SET

```python
# SET → collezione non ordinata senza duplicati

# Example: unique railway machines used in a project
machines = {"tamping machine", "ballast regulator", "tamping machine"}

print(machines)

# aggiungere una macchina
machines.add("rail grinder")
print(machines)

# verifica presenza
print("tamping machine" in machines)
```

# DICTIONARY

```python
# DICTIONARY → struttura chiave-valore (fondamentale per data analysis)

# Example: railway project information
project = {
    "name": "Milan-Verona Upgrade",
    "length_km": 120,
    "status": "in progress"
}

print(project)

# accesso tramite chiave
print(project["name"])

# aggiungere un nuovo campo
project["budget_million_eur"] = 250
print(project)
```

## REALISTICS DATA ANALYST EXAMPLE

```python
# esempio reale: lista di progetti ferroviari (tipico scenario data analyst)

projects = [
    {"name": "Line A Upgrade", "km": 50, "status": "completed"},
    {"name": "High Speed Line", "km": 200, "status": "in progress"},
    {"name": "Metro Extension", "km": 30, "status": "planned"}
]

# estrarre solo i progetti in corso
in_progress_projects = []

for p in projects:
    if p["status"] == "in progress":
        in_progress_projects.append(p)

print(in_progress_projects)
```
