# LIST

struttura ordinata e modificabile (utile per sequenze operative)

```python
# Example: list of railway maintenance tasks -> task = tarea
maintenance_tasks = ["track inspection", "ballast tamping", "rail alignment"]

print(maintenance_tasks)

# accesso al primo elemento
print(maintenance_tasks[0])

# aggiungere un nuovo task con il metodo .append()
maintenance_tasks.append("overhead line check")
print(maintenance_tasks)

# lunghezza della lista
print(len(maintenance_tasks))
```

> Alcuni metodi comuni dell`oggeto lista sono:
- .append() ---> agerga un elemento a la list
- .index() ---> restituye la posicion de un elemento
- .insert() ---> inserta un objeto en una poscion especifica
- .count() ---> cuantas veces un elemento aparece en la list (porque se permiten los duplicados)
- .len() ---> cuenta los elementos de una lista
- .remove() ---> remueve un elemento especifico
- .pop() ---> remueve o sustituye un elemento por default el ultimo (solucion lenta, mejor usar 'deque')
- .sort() ---> ordena la lista de forma creciente (modifica la lista original)
- .reverse() ---> invierte el orde de los elementos
- .copy() ---> crea una copia de la lista
- .clear() ---> vacia completamente la lista

```python
# lista di tratte ferroviarie
railway_lines = ["Milano-Brescia", "Brescia-Verona", "Verona-Vicenza"]

# numero di tratte
num_lines = len(railway_lines)  # conta gli elementi della lista

print(num_lines)  # output: 3
```

```python
my_list = []
my_list.append('pay bills')
my_list.append('tidy up')
my_list.append('walk the dog')
my_list.append('cook dinner')
print(my_list)
```

```python
# si quisiera agregar un elmento a una posicion puntual dentro de mi lista, primero debo utilizar el metodo .index() para determinar el inidice el elemento antes del que quiero instertar el nuevo elemento
i=my_list.index('cook dinner')
# esto se conbertira en el indice de la nueva actividad que ahora puede ser agregada utilizando el metodo .insert()
my_list.insert(i, 'go to the pharmacy')
print(my_list)
```

## Lista come coda (deque)
usar `collections.deque`, es mas optimizado per code FIFO, uso `deque` quando:
- ho molete `insert`/`remove` all'inizio
- lavoro son stream di dati ()
- gestisco code di eventi (logs, sensori, interventi)


> log = registro cronologico di enventi generati da un sistema. sono la materia prima dei dati e arrivano **in tempo reale**

> quando uso `pandas` no uso `deque`

> `FIFO` = first-in, first-out   `LIFO` = last-in, first-out

- servono per:
    - monitoraggio sistema
    - analisis problemi
    - clacolo KPI
    - AUDIT operativo

- esempi di log reali
    - passaggio treno su una tratta
    - ritardo registrato
    - errore segnalamento
    - intervento manutenzione
    - stato macchinario

>  Sintesi (mentalità senior)

- log = flusso continuo di eventi
- deque = gestione efficiente del flusso
- pandas = analisi del flusso

```python
# log eventi ferroviari (simulazione)
logs = [
    {"timestamp": "2026-04-10 08:00", "event": "Train passed", "line": "Milano-Brescia"},  # passaggio treno
    {"timestamp": "2026-04-10 08:05", "event": "Delay detected", "delay": 5},             # ritardo
    {"timestamp": "2026-04-10 08:10", "event": "Signal failure"}                          # guasto
]

print(logs)
```

### Coda FIFO (più comune)

```python
from collections import deque  # importo deque

# coda interventi ferroviari
queue = deque()

queue.append("Inspection Milano-Brescia")   # aggiunge in fondo
queue.append("Maintenance Brescia-Verona")

task = queue.popleft()  # rimuove il primo elemento (O(1))

print(task)
```

### Buffer di envti (streaming)

```python
from collections import deque #importo deque

# ultimi 3 eventi di linea
events = deque(maxlen=3) # buffer limitato a 3 eventi

events.append('train delay 5 min') # aggiungo evento
events.append('siganl failure') # aggiungo envento
events.append('track inspection') # aggiungo evento
events.append('maintenance complete') # si come ho messo un max di 3 allora il primo vienen eliminato automaticamente

print(events)
```

### Stack + Queue ibrido

```python
from collections import deque # importo deque

tasks = deque()

tasks.append('task a') # push
tasks.append('task b')

tasks.pop() # LIFO
tasks.popleft() # LIFO
```

>  Mental model (importante)

- `deque` = prima dei dati (stream / pipeline)
- `pandas` = dopo (analisi / reporting)

### LISTA COME PILA (stack)

Una pila implementa la metodologia `LIFO` in cui l'ultimo elemento aggiunto e il primo elemento recuperato.
- `.append()` --> push (inserire)
- `.pop()` --> pop (rimuovere)

```python
# pila di interventi ferrroviari urgenti
stack = []

# con push aggiungo interventi
stack.append('inspection milano-brescia') # inserisco intervento
stack.append('signal check verona') # inserisco intervento
stack.append('track repari vincenza') # inserisco intervento

# pop (il ultimo intervento viene eseguito per primo)
last_tastk = stack.pop() # rimuovo il ultimo elemento (LIFO)

print(last_tastk)
```

> Quando usarlo

quando gestisci:
- operazioni urgenti
- undo/redo
- processi a priorità immediata

# TUPLE

Una `tuple` e una struttura dati che rappresenta un insieme fisso di valori correlati, ha un ordine, contiene piu valori e non puo cambiare

```python
# coordinate di un punto sulla linea ferroviaria
track_point = (45.5416, 10.2118) #latitudine, logitudine

# accesso ai valori
lat = track_point[0] # leggo latitudine
lon = track_point[1] # leggo longitudine

print(lat,lon)
```

### Immutabilità
Una volta creata non puo essere modificata.

```python
track_point[0] = 50 # TypeError
```

> Perche usare TUPLA se i dati (dei sensori) cambiano?

si, i dati cambiano nel tempo, ma il **singolo record NON deve cambiare**
- ogni lettura del sensore = un nuovo record
- quel record = **immutabile**
si crea una nuova **tupla** per ogni rilevazione

> Non possso modificare o eliminiare `elementi interni` pero si posso eliminare il record / la tupla intera

```python
# tupla
reading = ('milano-brescia', 5, 'ok')
# no posso eliminare elementi della tupla
del reading[1] # errore
```

```python
# tupla
reading = ('milano-brescia', 5, 'ok')
# posso eliminare la variabile
del reading
print(reading)
```

### Lista di tuple

```python
# lista di rilevazioni
sensor_data = [
    ('milano-brescia', 5, 'ok'),
    ('brescia-verona', 0, 'ok'),
    ('verona-vincenza', 12, 'delay')
]
# rimuovo una tupla specifica
sensor_data.remove(('brescia-verona', 0, 'ok')) # rimuovo record

print(sensor_data)
```

### Quando usarla
Per garantire la integrita dei dati, quando i dati non devono cambiare, esempi reali:
- coordinate GPS
- ID fissi
- configurazioni
- record "read-only"

```python
# simulazione dati provenienti da sistema ferroviario

def get_sensor_data():
    #dati generati automaticamente dal sistema
    return ('milano-brescia', 5, 'ok') # linea, delay, stato

sensor_readings = []

# simulazione acquisizione continua
for _ in range(3):
    reading = get_sensor_data() # ricervo dato automatico
    sensor_readings.append(reading) # salvo storico
    
print(sensor_readings)
```

### Aggiornare una tupla

- Non modifico la tupla
- creo una nuova tupla aggiornata

Pattern corretto
- non rompo immutabilita
- mantengo storico corretto

```python
# record originale
reading = ("Milano-Brescia", 5, "OK")  # linea, delay, stato

# creo nuovo record aggiornato
updated_reading = (reading[0], 10, "DELAY")  # aggiorno delay e stato

print(updated_reading)
```

Caso reale (lista di record)

```python
# lista rilevazioni
sensor_data = [
    ("Milano-Brescia", 5, "OK"),
    ("Brescia-Verona", 0, "OK")
]

# aggiorno il primo record
old = sensor_data[0]  # prendo record originale

new = (old[0], 8, "DELAY")  # creo nuovo record aggiornato

sensor_data[0] = new  # sostituisco nella lista

print(sensor_data)
```

# DICTIONARY

Collexioni mutabili e NON ORDINATE di coppie `chiavi-valore`

Un dizionario e deliminitato da parentesi graffe {}. i dizionari hanno il vantaggio di assegnare un'etichetta a ogni dato

`{'Marchio': 'Ford', 'Modello': 'Mustang', 'Anno': '1964'}`


```python
# essempio da 'pithon per data science' - yuli vasiliev
car = {
    "brand": "Volkswagen",
    "style": "Sedan",
    "model": "Jetta"
}

print(car.setdefault("anno", "2002")) # .setdefault() inserisce la coppia chiave-valore e restituisce il valore
```

ogni dato e accessibile tramite una chiave unica

```python
# dati di un intervento ferroviario
intervention = {
    "line": "milano-brescia",       #linea interessata
    "delay": 5,                     # ritardo in minuti
    "status": "in_progress"         # stato del intervento
}

# accesso ai dati
print(intervention["line"])
```

### Quando usarla

usar `dict` quando ho dati strutturati con significato per esempio:
- log intervento
- evento sensore
- report manutenzione

```python
# evento sensore ferroviario
event = {
    "timestamp": "2026-04-10 08:00",     # tempo evento
    "line": "milano-brescia",            # linea
    "eventy_type": "delay",               # tipo evento
    "value": 5                           # valore misurato      
}

print(event)
```

### Perche usare dict

rispetto a tuple:
- piu leggibile
- piu flessibile
- piu vicino a JSON / database

```python
# operazioni base

# aggiungere campo
event["machine"] = "rincalzatrice" #aggiungo info

# modificare valore
event["value"] = 7 # aggiorno ritardo

# iterare
for key, value in event.items():
    print(key, value)   # stampo chiave-valore
```

# SET

Collezione NON ORDINATA di elementi unici (non duplicati)

```python
# set per rimuovere duplicati dalle sequienza
lst = ['jhon silver', 'tim gemison', 'jhon silver', 'maya smith']
lst = list(set(lst))
print(lst)
```

```python
# macchinari utilizzati nei cantieri
machines = {'tamping machine', 'ballast regulator', 'tamping machine'}  # duplicato

print(machines) # il duplicato viene eliminato automaticamente
```

### Quando usarlo

- per eliminare duplicate
- per verificare appartenenza
- per fare operaoji tra insiemi

```python
# macchinari utilizzati IN DUE CANTIERI
site_A = {'tamping machine', 'ballast regulator', 'welding machine'}    #cantiere A
site_B = {'tamping machine', 'inspection vehio'}

# machinari in comun
common = site_A.intersection(site_B)    # elementi condivivisi

print(common)
```

### Operaczione principale

```python
# unione (tutti i macchinari)
union = site_A.union(site_B)

# differenza (solo A)
diff = site_A.difference(site_B)

print(union)
print(diff)
```
