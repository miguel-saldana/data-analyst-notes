# Le strutture dati di Python

| Data Type  | Ordered | Mutable | Allows Duplicates | Indexing | Key-Value Structure | Syntax Example |
|-----------|--------|--------|------------------|----------|--------------------|----------------|
| List      | Yes    | Yes    | Yes              | Yes      | No                  | [1, 2, 3]      |
| Tuple     | Yes    | No     | Yes              | Yes      | No                 | (1, 2, 3)      |
| Dictionary| Yes*   | Yes    | No (keys)        | No       | Yes                | {"a": 1, "b": 2} |
| Set       | No     | Yes    | No               | No       | No                 | {1, 2, 3}      |

> *Dictionaries are ordered as of Python 3.7+, but they are not indexed like lists or tuples.
